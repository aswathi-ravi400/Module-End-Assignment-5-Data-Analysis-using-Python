■ Handle using: dropna(), fillna(), median/mode, forward/backward-fill

```python
df['gender']=df['gender'].fillna('Unknown')
df['sentiment']=df['sentiment'].fillna(df['sentiment'].mode()[0])
```

```python
df.isnull().sum()
```

Explanation:Used fillna() and the median of each numerical column to replace missing values in age,likes,comments and shares and decided not to use dropna() on the main dataset because it removes rows containing missing values,which can lead to loss of useful information.Decided not to use forward fill because it replaces missing values with the previous row's value. In this social media dataset,the previous row's value may not be appropriate for the missing observation.Decided not to use backward fill because it replaces missing values with the next row's value.The next row may represent a different user or social media post,so copying its value may produce inaccurate information.Missing values in the gender column were replaced with "Unknown" using fillna().Missing values in the sentiment column were replaced with the mode,which is the most frequently occurring value.This approach preserves all rows while handling missing categorical data.After handling missing data rechecked if there is any missing values left behind.
